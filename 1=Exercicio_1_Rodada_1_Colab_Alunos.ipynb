## AI-Driven Software Engineering — Testes Automatizados com IA

### Rodada 1 — A IA conhece apenas o código

Neste exercício, você vai investigar uma pergunta que parece simples, mas é central para o uso profissional de Inteligência Artificial em testes de software:

> **Se uma IA gerar uma suíte de testes e todos os testes passarem, isso significa que o software está correto?**

Você receberá uma pequena implementação Python e utilizará uma ferramenta de IA para produzir testes automatizados.

**Nesta primeira rodada, você não receberá a especificação funcional.**

Isso é proposital.

O objetivo não é descobrir quem consegue “achar o bug” mais rápido. O objetivo é observar **o que realmente pode ser concluído quando a única fonte disponível é o próprio código que está sendo testado**.

## 🎯 Objetivos de aprendizagem

Ao final desta rodada, você deverá ser capaz de:

- distinguir **comportamento observado** de **comportamento correto**;
- identificar de onde veio o resultado esperado de cada teste;
- perceber que uma suíte totalmente verde pode ainda fornecer evidência insuficiente sobre a correção do software;
- usar IA para gerar testes sem tratar automaticamente a resposta do modelo como verdade;
- diferenciar uma **hipótese levantada pela IA** de uma **regra de negócio conhecida**;
- registrar evidências antes de tirar conclusões.

> **Importante:** não tente adivinhar a regra de negócio. Trabalhe apenas com as informações realmente disponíveis nesta rodada.

1. **Não altere a função `calcular_frete`.**
2. Nesta rodada, considere que **não existe especificação funcional disponível**.
3. Você pode usar uma ferramenta de IA autorizada pela professora.
4. A IA pode analisar o código e sugerir testes.
5. Você deve executar os testes gerados — não basta confiar no código produzido pelo modelo.
6. Para cada teste, registre **de onde veio a expectativa utilizada no `assert`**.
7. Se a IA levantar uma possível regra de negócio, trate-a como **hipótese**, não como fato.
8. Não procure na internet uma especificação para a função. O comportamento de negócio será revelado somente em outra etapa do exercício.

- modificar o código para “melhorá-lo”;

````python
import pytest
from frete import calcular_frete

def test_frete_gratis_premium_acima_200():
    # Expectativa (0.0): Inferida da condição 'if cliente_premium and valor_compra > 200'
    assert calcular_frete(200.01, True) == 0.0

def test_frete_padrao_premium_exatamente_200():
    # Expectativa (20.0): A condição '> 200' é falsa para 200.0, caindo no 'return 20.0'
    assert calcular_frete(200.0, True) == 20.0

def test_frete_padrao_premium_abaixo_200():
    # Expectativa (20.0): A condição '> 200' é falsa, caindo no 'return 20.0'
    assert calcular_frete(199.99, True) == 20.0

def test_frete_padrao_nao_premium_acima_200():
    # Expectativa (20.0): A condição 'cliente_premium' é False, caindo no 'return 20.0'
    assert calcular_frete(250.0, False) == 20.0

def test_frete_padrao_limite_zero():
    # Expectativa (20.0): O valor 0.0 falha no teste de erro (< 0) e cai no 'return 20.0'
    assert calcular_frete(0.0, False) == 20.0

def test_erro_valor_negativo():
    # Expectativa (ValueError): Inferida da condição inicial 'if valor_compra < 0'
    with pytest.raises(ValueError, match="valor_compra não pode ser negativo"):
        calcular_frete(-0.01, True)
````
